```python
# prompt: Gere agora um K-Means, algoritmo de aprendizado não supervisionado

import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import pandas as pd # Import pandas as it's needed for DataFrame operations

# Importando dataset
grupo = pd.read_csv('marketing_campaign2.csv')

# Traduzindo os nomes das colunas para português
colunas_originais = ['Education', 'Income', 'Kidhome', 'Teenhome', 'Recency', 'MntWines',
                     'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts',
                     'MntGoldProds', 'NumDealsPurchases', 'NumWebPurchases', 'NumCatalogPurchases',
                     'NumStorePurchases', 'NumWebVisitsMonth', 'Customer_For', 'Age', 'Spent',
                     'Living_With', 'Children', 'Family_Size', 'Is_Parent']

colunas_traduzidas = ['Educacao', 'Renda', 'CriancasEmCasa', 'AdolescentesEmCasa', 'Recencia', 'ValorVinho',
                      'ValorFruta', 'ValorCarne', 'ValorPeixe', 'ValorDoce',
                      'ValorOuro', 'NumComprasPromocao', 'NumComprasWeb', 'NumComprasCatalogo',
                      'NumComprasLoja', 'NumVisitasWebMes', 'ClienteDesde', 'Idade', 'Gasto',
                      'MorandoCom', 'Filhos', 'TamanhoFamilia', 'EhPaiMae']

grupo.columns = colunas_traduzidas

# Tratamento de valores ausentes (se houver)
# Para o K-Means, não podemos ter valores nulos. Vamos preencher com a média.
grupo['Renda'].fillna(grupo['Renda'].mean(), inplace=True)

# Seleção das features para o K-Means
# Vamos usar as variáveis numéricas relacionadas a gastos e compras, e a renda.
features_para_kmeans = ['Renda', 'ValorVinho', 'ValorFruta', 'ValorCarne', 'ValorPeixe',
                        'ValorDoce', 'ValorOuro', 'NumComprasPromocao', 'NumComprasWeb',
                        'NumComprasCatalogo', 'NumComprasLoja', 'NumVisitasWebMes', 'Gasto']

X = grupo[features_para_kmeans]

# Escalonamento dos dados
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Determinando o número ideal de clusters (usando o método do cotovelo e Silhouette Score)

# Método do Cotovelo
wcss = [] # Within-Cluster Sum of Squares
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Método do Cotovelo')
plt.xlabel('Número de Clusters')
plt.ylabel('WCSS')
plt.show()

# Silhouette Score
# Calcula o Silhouette Score para diferentes números de clusters
silhouette_scores = []
for n_clusters in range(2, 11):
    kmeans = KMeans(n_clusters=n_clusters, init='k-means++', random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(X_scaled)
    silhouette_avg = silhouette_score(X_scaled, cluster_labels)
    silhouette_scores.append(silhouette_avg)
    print(f"Para n_clusters = {n_clusters}, o Silhouette Score é {silhouette_avg}")

plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), silhouette_scores, marker='o', linestyle='--')
plt.title('Silhouette Score por Número de Clusters')
plt.xlabel('Número de Clusters')
plt.ylabel('Silhouette Score')
plt.show()

# Baseado nos gráficos acima (método do cotovelo e silhouette score), escolha um número adequado de clusters.
# Por exemplo, se o cotovelo estiver claro em 3 ou 4, e o silhouette score for razoavelmente alto para esse número.
# Vamos escolher um número de clusters para demonstração, por exemplo, 3.
n_clusters_escolhido = 3

# Aplicando o K-Means com o número de clusters escolhido
kmeans_final = KMeans(n_clusters=n_clusters_escolhido, init='k-means++', random_state=42, n_init=10)
grupo['Cluster'] = kmeans_final.fit_predict(X_scaled)

# Análise dos clusters
print("\nTamanho de cada cluster:")
print(grupo['Cluster'].value_counts())

# Características dos clusters (média das features para cada cluster)
print("\nCaracterísticas dos Clusters (Médias):")
print(grupo.groupby('Cluster')[features_para_kmeans].mean())

# Visualização dos clusters (exemplo com Renda e Gasto)
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Renda', y='Gasto', hue='Cluster', data=grupo, palette='viridis', s=100)
plt.title('Clusters de Clientes (Renda vs Gasto)')
plt.xlabel('Renda')
plt.ylabel('Gasto')
plt.show()

# Exemplo de visualização com outras variáveis (substitua 'ValorVinho' e 'ValorCarne' por outras features_para_kmeans)
# plt.figure(figsize=(10, 6))
# sns.scatterplot(x='ValorVinho', y='ValorCarne', hue='Cluster', data=grupo, palette='viridis', s=100)
# plt.title('Clusters de Clientes (Valor Vinho vs Valor Carne)')
# plt.xlabel('Valor Vinho')
# plt.ylabel('Valor Carne')
# plt.show()
```
